```python
# ============================================================
# BLOCK 1: Wood Chip UDA — Final Model (noOrth_ent_noPsl)
# Based on ablation Exp 6: best target accuracy (0.820)
# Active losses: loss_cls + loss_dc + loss_dh + loss_ent
# Removed:       loss_orth, loss_pseudo
# ============================================================

# ============================================================
# SECTION 0: Imports & Config
# ============================================================
import os
import sys
import copy
import math
import random
from collections import deque
import numpy as np
import pandas as pd
from PIL import Image
from tqdm import tqdm
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

sys.path.append(r"D:\Woodchip_moisture_content\Dataset")
from dataset_1_3 import SourceDataset, TargetDataset, WoodChipTargetEvalDataset

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.deterministic = False
torch.backends.cudnn.benchmark = True

SAVE_DIR = r"D:\Woodchip_moisture_content\Final\Final_model_output\1_3"
os.makedirs(SAVE_DIR, exist_ok=True)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
USE_AMP = torch.cuda.is_available()

BACKBONE_NAME = "convnext_small"
NUM_CLASSES = 3
FEATURE_DIM = 512
BATCH_SIZE = 16
NUM_EPOCHS = 30
NUM_WORKERS = 0
DROPOUT = 0.2
BACKBONE_LR = 1e-5
HEAD_LR = 1e-4
WEIGHT_DECAY = 1e-4
LAMBDA_COMMON = 0.5
LAMBDA_HETERO = 0.1
LAMBDA_COMMON_MIN = 0.05
LAMBDA_COMMON_MAX = 1.00
FREEZE_BACKBONE_EPOCHS = 2
DBS_MA_WINDOW = 3
UDA_ALPHA = 10.0
FOCAL_GAMMA = 2.0
LAMBDA_ENT = 0.1

LABEL_NAMES = ['Dry', 'Medium', 'Wet']

print(f"Device    : {DEVICE}")
print(f"Backbone  : {BACKBONE_NAME}")
print(f"Outputs   : {SAVE_DIR}")
print(f"Epochs    : {NUM_EPOCHS}")
print(f"Focal gamma     : {FOCAL_GAMMA}")
print(f"Entropy lambda  : {LAMBDA_ENT}")
print(f"Orth loss       : DISABLED")
print(f"Pseudo-labeling : DISABLED")

# ============================================================
# SECTION 1: Dataset Setup
# ============================================================
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD  = [0.229, 0.224, 0.225]

train_transform = transforms.Compose([
    transforms.Resize((256, 256)),
    transforms.RandomResizedCrop(224, scale=(0.75, 1.0), ratio=(0.85, 1.15)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomVerticalFlip(),
    transforms.RandomRotation(15),
    transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.2, hue=0.05),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
])

eval_transform = transforms.Compose([
    transforms.Resize((256, 256)),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
])

class TransformWrapper(Dataset):
    def __init__(self, base_dataset, transform):
        self.samples  = base_dataset.samples
        self.transform = transform
        self.labeled  = isinstance(self.samples[0], tuple)

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        if self.labeled:
            path, label = self.samples[idx]
            img = Image.open(path).convert('RGB')
            return self.transform(img), label
        else:
            path = self.samples[idx]
            img = Image.open(path).convert('RGB')
            return self.transform(img)

src_train_ds = TransformWrapper(SourceDataset(), train_transform)
src_eval_ds  = TransformWrapper(SourceDataset(), eval_transform)
tgt_train_ds = TransformWrapper(TargetDataset(), train_transform)
tgt_eval_ds  = TransformWrapper(WoodChipTargetEvalDataset(), eval_transform)

src_loader      = DataLoader(src_train_ds, batch_size=BATCH_SIZE, shuffle=True,
                             num_workers=NUM_WORKERS, drop_last=True, pin_memory=True)
tgt_loader      = DataLoader(tgt_train_ds, batch_size=BATCH_SIZE, shuffle=True,
                             num_workers=NUM_WORKERS, drop_last=True, pin_memory=True)
src_eval_loader = DataLoader(src_eval_ds, batch_size=BATCH_SIZE, shuffle=False,
                             num_workers=NUM_WORKERS, pin_memory=True)
tgt_eval_loader = DataLoader(tgt_eval_ds, batch_size=BATCH_SIZE, shuffle=False,
                             num_workers=NUM_WORKERS, pin_memory=True)

print(f"\nSource train : {len(src_train_ds)} images (labeled)")
print(f"Target train : {len(tgt_train_ds)} images (no labels used)")
print(f"Target eval  : {len(tgt_eval_ds)} images")

# ============================================================
# SECTION 2: Model Architecture
# ============================================================
class GradReverse(torch.autograd.Function):
    @staticmethod
    def forward(ctx, x, lambd):
        ctx.lambd = lambd
        return x.view_as(x)

    @staticmethod
    def backward(ctx, grad_output):
        return -ctx.lambd * grad_output, None

def grad_reverse(x, lambd=1.0):
    return GradReverse.apply(x, lambd)

class AdvancedBackbone(nn.Module):
    def __init__(self, backbone_name="convnext_small", pretrained=True):
        super().__init__()
        self.backbone_name = backbone_name.lower()
        if self.backbone_name == "convnext_tiny":
            weights = models.ConvNeXt_Tiny_Weights.IMAGENET1K_V1 if pretrained else None
            b = models.convnext_tiny(weights=weights)
            self.features = b.features
            self.pool = nn.AdaptiveAvgPool2d((1, 1))
            self.feature_dim = b.classifier[2].in_features
        elif self.backbone_name == "convnext_small":
            weights = models.ConvNeXt_Small_Weights.IMAGENET1K_V1 if pretrained else None
            b = models.convnext_small(weights=weights)
            self.features = b.features
            self.pool = nn.AdaptiveAvgPool2d((1, 1))
            self.feature_dim = b.classifier[2].in_features
        elif self.backbone_name == "efficientnet_v2_s":
            weights = models.EfficientNet_V2_S_Weights.IMAGENET1K_V1 if pretrained else None
            b = models.efficientnet_v2_s(weights=weights)
            self.features = b.features
            self.pool = nn.AdaptiveAvgPool2d((1, 1))
            self.feature_dim = b.classifier[1].in_features
        elif self.backbone_name == "swin_t":
            weights = models.Swin_T_Weights.IMAGENET1K_V1 if pretrained else None
            b = models.swin_t(weights=weights)
            self.features = b.features
            self.norm = b.norm
            self.permute = b.permute
            self.pool = nn.AdaptiveAvgPool2d((1, 1))
            self.feature_dim = b.head.in_features
        elif self.backbone_name == "resnet50":
            weights = models.ResNet50_Weights.IMAGENET1K_V2 if pretrained else None
            b = models.resnet50(weights=weights)
            self.features = nn.Sequential(*list(b.children())[:-1])
            self.feature_dim = b.fc.in_features
        else:
            raise ValueError(f"Unknown backbone: {backbone_name}")

    def forward(self, x):
        if self.backbone_name in ["convnext_tiny", "convnext_small", "efficientnet_v2_s"]:
            x = self.features(x)
            x = self.pool(x)
            return torch.flatten(x, 1)
        elif self.backbone_name == "swin_t":
            x = self.features(x)
            x = self.norm(x)
            x = self.permute(x)
            x = self.pool(x)
            return torch.flatten(x, 1)
        elif self.backbone_name == "resnet50":
            return torch.flatten(self.features(x), 1)

class WoodChipUDA(nn.Module):
    def __init__(self, backbone_name="convnext_small", feature_dim=512,
                 num_classes=3, dropout=0.2, pretrained=True):
        super().__init__()
        self.backbone = AdvancedBackbone(backbone_name, pretrained)
        in_dim = self.backbone.feature_dim

        self.common_encoder = nn.Sequential(
            nn.Linear(in_dim, feature_dim),
            nn.LayerNorm(feature_dim),
            nn.GELU(),
            nn.Dropout(dropout)
        )
        self.hetero_encoder = nn.Sequential(
            nn.Linear(in_dim, feature_dim),
            nn.LayerNorm(feature_dim),
            nn.GELU(),
            nn.Dropout(dropout)
        )
        self.classifier = nn.Linear(feature_dim, num_classes)
        self.domain_common = nn.Sequential(
            nn.Linear(feature_dim, 256),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(256, 2)
        )
        self.domain_hetero = nn.Sequential(
            nn.Linear(feature_dim, 256),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(256, 2)
        )

    def forward(self, x, grl_lambda=1.0):
        f = self.backbone(x)
        zc = self.common_encoder(f)
        zh = self.hetero_encoder(f)
        y_logits        = self.classifier(zc)
        d_common_logits = self.domain_common(grad_reverse(zc, grl_lambda))
        d_hetero_logits = self.domain_hetero(zh)
        return y_logits, d_common_logits, d_hetero_logits, zc, zh

# ============================================================
# SECTION 3: Loss Functions
# ============================================================
class FocalLoss(nn.Module):
    def __init__(self, gamma=2.0):
        super().__init__()
        self.gamma = gamma

    def forward(self, logits, targets):
        ce_loss = F.cross_entropy(logits, targets, reduction='none')
        pt = torch.exp(-ce_loss)
        return ((1.0 - pt) ** self.gamma * ce_loss).mean()

focal_loss = FocalLoss(gamma=FOCAL_GAMMA)

def entropy_loss(logits):
    probs     = F.softmax(logits, dim=1)
    log_probs = F.log_softmax(logits, dim=1)
    return -torch.sum(probs * log_probs, dim=1).mean()

def compute_dbs(common_acc, hetero_acc):
    return hetero_acc - abs(common_acc - 50.0)

def compute_uda_score(dbs_ma, cls_loss, alpha=10.0):
    return dbs_ma - alpha * cls_loss

# ============================================================
# SECTION 4: Evaluation Helpers
# ============================================================
@torch.no_grad()
def evaluate_target(model, loader, device, use_amp):
    model.eval()
    correct, total = 0, 0
    for x, y in loader:
        x, y = x.to(device), y.to(device)
        with torch.amp.autocast("cuda", enabled=use_amp):
            y_logits, _, _, _, _ = model(x, grl_lambda=0.0)
        correct += (y_logits.argmax(1) == y).sum().item()
        total   += y.size(0)
    return 100.0 * correct / total

@torch.no_grad()
def evaluate_domain_separation(model, src_loader, tgt_loader, device, use_amp):
    model.eval()
    common_correct = hetero_correct = total = 0
    src_iter = iter(src_loader)
    tgt_iter = iter(tgt_loader)
    for _ in range(min(len(src_loader), len(tgt_loader))):
        xs, _ = next(src_iter)
        xt = next(tgt_iter)
        if isinstance(xt, (list, tuple)):
            xt = xt[0]
        x = torch.cat([xs.to(device), xt.to(device)], dim=0)
        domain_labels = torch.cat([
            torch.zeros(xs.size(0), dtype=torch.long),
            torch.ones(xt.size(0),  dtype=torch.long)
        ]).to(device)
        with torch.amp.autocast("cuda", enabled=use_amp):
            _, d_common, d_hetero, _, _ = model(x, grl_lambda=0.0)
        common_correct += (d_common.argmax(1) == domain_labels).sum().item()
        hetero_correct += (d_hetero.argmax(1) == domain_labels).sum().item()
        total          += domain_labels.size(0)
    return 100.0 * common_correct / total, 100.0 * hetero_correct / total

def run_full_evaluation(model, loader, device, use_amp):
    model.eval()
    all_preds, all_probs, all_labels, all_imgs, all_feats = [], [], [], [], []
    with torch.no_grad():
        for imgs, labels in loader:
            imgs = imgs.to(device)
            with torch.amp.autocast("cuda", enabled=use_amp):
                y_logits, _, _, zc, _ = model(imgs, grl_lambda=0.0)
            probs = F.softmax(y_logits, dim=1)
            preds = y_logits.argmax(dim=1)
            all_preds.extend(preds.cpu().numpy())
            all_probs.extend(probs.cpu().numpy())
            all_labels.extend(labels.numpy())
            all_imgs.extend(imgs.cpu().numpy().transpose(0, 2, 3, 1))
            all_feats.extend(zc.cpu().numpy())
    return (np.array(all_preds), np.array(all_probs),
            np.array(all_labels), np.array(all_imgs), np.array(all_feats))

def set_backbone_trainable(model, trainable):
    for p in model.backbone.parameters():
        p.requires_grad = trainable

# ============================================================
# SECTION 5: Model & Optimizer
# ============================================================
model = WoodChipUDA(
    backbone_name=BACKBONE_NAME,
    feature_dim=FEATURE_DIM,
    num_classes=NUM_CLASSES,
    dropout=DROPOUT,
    pretrained=True
).to(DEVICE)

print(f"\nBackbone     : {BACKBONE_NAME}")
print(f"Backbone dim : {model.backbone.feature_dim}")

optimizer = torch.optim.AdamW([
    {"params": model.backbone.parameters(),       "lr": BACKBONE_LR},
    {"params": model.common_encoder.parameters(), "lr": HEAD_LR},
    {"params": model.hetero_encoder.parameters(), "lr": HEAD_LR},
    {"params": model.classifier.parameters(),     "lr": HEAD_LR},
    {"params": model.domain_common.parameters(),  "lr": HEAD_LR},
    {"params": model.domain_hetero.parameters(),  "lr": HEAD_LR},
], weight_decay=WEIGHT_DECAY)

scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
    optimizer, T_max=NUM_EPOCHS, eta_min=1e-6
)
scaler = torch.amp.GradScaler("cuda", enabled=USE_AMP)

# ============================================================
# SECTION 6: Training Loop
# ============================================================
dbs_queue       = deque(maxlen=DBS_MA_WINDOW)
best_target_acc = 0.0
best_raw_dbs    = -1e9
best_smooth_dbs = -1e9
best_uda_score  = -1e9
lambda_common   = LAMBDA_COMMON
history         = []

print("\n" + "=" * 70)
print("TRAINING")
print("=" * 70)

for epoch in range(1, NUM_EPOCHS + 1):
    set_backbone_trainable(model, epoch > FREEZE_BACKBONE_EPOCHS)
    model.train()

    src_iter = iter(src_loader)
    tgt_iter = iter(tgt_loader)
    num_iter = min(len(src_loader), len(tgt_loader))

    ep_loss = ep_cls = ep_dc = ep_dh = ep_ent = 0.0

    p          = epoch / NUM_EPOCHS
    grl_lambda = 0.5 * (2.0 / (1.0 + math.exp(-10 * p)) - 1.0)

    for _ in tqdm(range(num_iter), desc=f"Epoch {epoch}/{NUM_EPOCHS}", leave=False):
        xs, ys = next(src_iter)
        xt = next(tgt_iter)
        if isinstance(xt, (list, tuple)):
            xt = xt[0]

        xs = xs.to(DEVICE, non_blocking=True)
        ys = ys.to(DEVICE, non_blocking=True)
        xt = xt.to(DEVICE, non_blocking=True)

        x = torch.cat([xs, xt], dim=0)
        domain_labels = torch.cat([
            torch.zeros(xs.size(0), dtype=torch.long),
            torch.ones(xt.size(0),  dtype=torch.long)
        ]).to(DEVICE)

        optimizer.zero_grad(set_to_none=True)

        with torch.amp.autocast("cuda", enabled=USE_AMP):
            y_logits, d_common, d_hetero, zc, zh = model(x, grl_lambda)

            src_logits = y_logits[:xs.size(0)]
            tgt_logits = y_logits[xs.size(0):]

            loss_cls = focal_loss(src_logits, ys)
            loss_dc  = F.cross_entropy(d_common, domain_labels)
            loss_dh  = F.cross_entropy(d_hetero, domain_labels)
            loss_ent = entropy_loss(tgt_logits)

            loss = (loss_cls
                    + lambda_common * loss_dc
                    + LAMBDA_HETERO  * loss_dh
                    + LAMBDA_ENT     * loss_ent)

        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=5.0)
        scaler.step(optimizer)
        scaler.update()

        ep_loss += loss.item()
        ep_cls  += loss_cls.item()
        ep_dc   += loss_dc.item()
        ep_dh   += loss_dh.item()
        ep_ent  += loss_ent.item()

    scheduler.step()

    avg_loss = ep_loss / num_iter
    avg_cls  = ep_cls  / num_iter
    avg_dc   = ep_dc   / num_iter
    avg_dh   = ep_dh   / num_iter
    avg_ent  = ep_ent  / num_iter

    target_acc              = evaluate_target(model, tgt_eval_loader, DEVICE, USE_AMP)
    common_acc, hetero_acc  = evaluate_domain_separation(model, src_loader, tgt_loader, DEVICE, USE_AMP)
    raw_dbs                 = compute_dbs(common_acc, hetero_acc)
    dbs_queue.append(raw_dbs)
    smooth_dbs = sum(dbs_queue) / len(dbs_queue)
    uda_score  = compute_uda_score(smooth_dbs, avg_cls, UDA_ALPHA)

    if common_acc > 60.0:
        lambda_common = min(lambda_common * 1.05, LAMBDA_COMMON_MAX)
    elif common_acc < 40.0:
        lambda_common = max(lambda_common * 0.95, LAMBDA_COMMON_MIN)

    history.append({
        "epoch": epoch, "backbone": BACKBONE_NAME,
        "loss": avg_loss, "cls_loss": avg_cls,
        "domain_common_loss": avg_dc, "domain_hetero_loss": avg_dh,
        "ent_loss": avg_ent,
        "target_acc_tracking_only": target_acc,
        "common_domain_acc": common_acc, "hetero_domain_acc": hetero_acc,
        "raw_dbs": raw_dbs, "smooth_dbs_ma3": smooth_dbs,
        "uda_score": uda_score, "grl_lambda": grl_lambda,
        "lambda_common_next": lambda_common
    })

    print(
        f"Epoch [{epoch:02d}/{NUM_EPOCHS}] "
        f"Loss: {avg_loss:.4f} | Cls: {avg_cls:.4f} | "
        f"Dc: {avg_dc:.4f} | Dh: {avg_dh:.4f} | Ent: {avg_ent:.4f} | "
        f"Target Acc: {target_acc:.2f}% | "
        f"DBS: {raw_dbs:.2f} | UDA Score: {uda_score:.2f} | GRL: {grl_lambda:.3f}"
    )

    if target_acc > best_target_acc:
        best_target_acc = target_acc
        torch.save(copy.deepcopy(model.state_dict()),
                   os.path.join(SAVE_DIR, f"best_{BACKBONE_NAME}_by_target_acc.pth"))
    if raw_dbs > best_raw_dbs:
        best_raw_dbs = raw_dbs
        torch.save(copy.deepcopy(model.state_dict()),
                   os.path.join(SAVE_DIR, f"best_{BACKBONE_NAME}_by_raw_DBS.pth"))
    if smooth_dbs > best_smooth_dbs:
        best_smooth_dbs = smooth_dbs
        torch.save(copy.deepcopy(model.state_dict()),
                   os.path.join(SAVE_DIR, f"best_{BACKBONE_NAME}_by_smooth_DBS.pth"))
    if uda_score > best_uda_score:
        best_uda_score = uda_score
        torch.save(copy.deepcopy(model.state_dict()),
                   os.path.join(SAVE_DIR, f"best_{BACKBONE_NAME}_by_UDA_score.pth"))

    pd.DataFrame(history).to_csv(
        os.path.join(SAVE_DIR, f"training_log_{BACKBONE_NAME}.csv"), index=False
    )

print(f"\nBest target acc (tracking) : {best_target_acc:.2f}%")
print(f"Best raw DBS               : {best_raw_dbs:.2f}")
print(f"Best smooth DBS            : {best_smooth_dbs:.2f}")
print(f"Best UDA score             : {best_uda_score:.2f}")

# ============================================================
# SECTION 7: Final Evaluation (best model by UDA score)
# ============================================================
print("\nLoading BEST model by UDA score for final evaluation...")
best_model_path = os.path.join(SAVE_DIR, f"best_{BACKBONE_NAME}_by_UDA_score.pth")
model.load_state_dict(torch.load(best_model_path, map_location=DEVICE))
print(f"Loaded: {best_model_path}")

src_preds, src_probs, src_labels_ev, src_imgs_np, src_feats = \
    run_full_evaluation(model, src_eval_loader, DEVICE, USE_AMP)
tgt_preds, tgt_probs, tgt_labels_ev, tgt_imgs_np, tgt_feats = \
    run_full_evaluation(model, tgt_eval_loader, DEVICE, USE_AMP)

def compute_metrics(y_true, y_pred):
    return {
        "Accuracy":  accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred, average='weighted', zero_division=0),
        "Recall":    recall_score(y_true, y_pred, average='weighted', zero_division=0),
        "F1":        f1_score(y_true, y_pred, average='weighted', zero_division=0),
    }

src_metrics = compute_metrics(src_labels_ev, src_preds)
tgt_metrics = compute_metrics(tgt_labels_ev, tgt_preds)

print("\n" + "=" * 52)
print("METRICS SUMMARY (Best UDA Score Model)")
print("=" * 52)
print(f"{'Metric':<20} {'Source':>10} {'Target':>10}")
print("-" * 52)
for m in ["Accuracy", "Precision", "Recall", "F1"]:
    print(f"{m:<20} {src_metrics[m]:>10.3f} {tgt_metrics[m]:>10.3f}")
print("=" * 52)
print(f"\nBest target acc during training = {best_target_acc:.3f}%")

# ============================================================
# SECTION 8: Save All Outputs
# ============================================================
src_paths = np.array([s[0] for s in src_eval_ds.samples])
tgt_paths = np.array([s[0] for s in tgt_eval_ds.samples])

np.save(os.path.join(SAVE_DIR, "Xs.npy"),              src_imgs_np)
np.save(os.path.join(SAVE_DIR, "ys_int.npy"),          src_labels_ev)
np.save(os.path.join(SAVE_DIR, "Xt_eval.npy"),         tgt_imgs_np)
np.save(os.path.join(SAVE_DIR, "yt_int.npy"),          tgt_labels_ev)
np.save(os.path.join(SAVE_DIR, "src_preds_vis.npy"),   src_preds)
np.save(os.path.join(SAVE_DIR, "src_probs_vis.npy"),   src_probs)
np.save(os.path.join(SAVE_DIR, "tgt_preds_vis.npy"),   tgt_preds)
np.save(os.path.join(SAVE_DIR, "tgt_probs_vis.npy"),   tgt_probs)
np.save(os.path.join(SAVE_DIR, "src_paths.npy"),       src_paths)
np.save(os.path.join(SAVE_DIR, "tgt_paths.npy"),       tgt_paths)
np.save(os.path.join(SAVE_DIR, "src_feats.npy"),       src_feats)
np.save(os.path.join(SAVE_DIR, "tgt_feats.npy"),       tgt_feats)
np.save(os.path.join(SAVE_DIR, "training_history.npy"), history)

print(f"\nAll outputs saved to: {SAVE_DIR}")

# ============================================================
# BLOCK 2: Confusion Matrix, t-SNE & Prediction Visualization
# ============================================================
import shutil
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report
from sklearn.manifold import TSNE

matplotlib.rcParams['figure.dpi'] = 150

Xs          = np.load(os.path.join(SAVE_DIR, "Xs.npy"))
ys_int      = np.load(os.path.join(SAVE_DIR, "ys_int.npy"))
Xt_eval     = np.load(os.path.join(SAVE_DIR, "Xt_eval.npy"))
yt_int      = np.load(os.path.join(SAVE_DIR, "yt_int.npy"))
src_preds_vis = np.load(os.path.join(SAVE_DIR, "src_preds_vis.npy"))
src_probs_vis = np.load(os.path.join(SAVE_DIR, "src_probs_vis.npy"))
tgt_preds_vis = np.load(os.path.join(SAVE_DIR, "tgt_preds_vis.npy"))
tgt_probs_vis = np.load(os.path.join(SAVE_DIR, "tgt_probs_vis.npy"))
src_paths     = np.load(os.path.join(SAVE_DIR, "src_paths.npy"))
tgt_paths     = np.load(os.path.join(SAVE_DIR, "tgt_paths.npy"))
src_feats     = np.load(os.path.join(SAVE_DIR, "src_feats.npy"))
tgt_feats     = np.load(os.path.join(SAVE_DIR, "tgt_feats.npy"))

print("All outputs loaded successfully.")
print(f"Source: {len(Xs)} images | Target: {len(Xt_eval)} images")

def print_metrics(y_true, y_pred, domain):
    acc  = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred, average='weighted', zero_division=0)
    rec  = recall_score(y_true, y_pred, average='weighted', zero_division=0)
    f1   = f1_score(y_true, y_pred, average='weighted', zero_division=0)
    print(f"\n{domain} — Acc: {acc:.3f} | Prec: {prec:.3f} | Rec: {rec:.3f} | F1: {f1:.3f}")

print_metrics(ys_int, src_preds_vis, "Source Domain")
print_metrics(yt_int, tgt_preds_vis, "Target Domain")

print("\nPer-class report — Target Domain:")
print(classification_report(yt_int, tgt_preds_vis, target_names=LABEL_NAMES, zero_division=0))

# Confusion matrices
fig, axes = plt.subplots(1, 2, figsize=(13, 5), dpi=300)
for ax, y_true, y_pred, title in zip(
    axes,
    [ys_int, yt_int],
    [src_preds_vis, tgt_preds_vis],
    ["Source Domain (source_1)", "Target Domain (source_2)"]
):
    cm  = confusion_matrix(y_true, y_pred)
    acc = accuracy_score(y_true, y_pred)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=LABEL_NAMES, yticklabels=LABEL_NAMES,
                linewidths=0.5, linecolor='gray',
                annot_kws={'size': 13, 'weight': 'bold'}, ax=ax)
    ax.set_xlabel('Predicted Label', fontsize=11, labelpad=6)
    ax.set_ylabel('True Label', fontsize=11, labelpad=6)
    ax.set_title(f"{title}\nAccuracy: {acc:.1%}", fontsize=12, fontweight='bold', pad=10)

plt.suptitle('Wood Chip UDA — Confusion Matrices', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
cm_path = os.path.join(SAVE_DIR, "confusion_matrix_combined.jpg")
plt.savefig(cm_path, format='jpg', dpi=300, bbox_inches='tight')
plt.show()
print(f"Saved: {cm_path}")

# t-SNE
print("\nRunning t-SNE (may take ~30-60 seconds)...")
all_feats  = np.concatenate([src_feats, tgt_feats], axis=0)
all_labels = np.concatenate([ys_int, yt_int], axis=0)
tsne = TSNE(n_components=2, perplexity=40, random_state=42, max_iter=1000)
emb  = tsne.fit_transform(all_feats)
src_emb = emb[:len(src_feats)]
tgt_emb = emb[len(src_feats):]

CLASS_COLORS = ['#e74c3c', '#2ecc71', '#3498db']
fig, ax = plt.subplots(figsize=(8, 6), dpi=300)
for cls_idx, cls_name in enumerate(LABEL_NAMES):
    mask_s = ys_int == cls_idx
    mask_t = yt_int == cls_idx
    ax.scatter(src_emb[mask_s, 0], src_emb[mask_s, 1],
               c=CLASS_COLORS[cls_idx], marker='o', s=25, alpha=0.6,
               label=f'{cls_name} (Src)')
    ax.scatter(tgt_emb[mask_t, 0], tgt_emb[mask_t, 1],
               c=CLASS_COLORS[cls_idx], marker='+', s=60, alpha=0.85,
               label=f'{cls_name} (Tgt)')

ax.set_title('t-SNE — Common Encoder Features\n(Circle = Source, Cross = Target)',
             fontsize=13, fontweight='bold')
ax.set_xlabel('t-SNE dim 1', fontsize=11)
ax.set_ylabel('t-SNE dim 2', fontsize=11)
ax.legend(fontsize=9, ncol=2, loc='best')
plt.tight_layout()
tsne_path = os.path.join(SAVE_DIR, "tsne_common_features.jpg")
plt.savefig(tsne_path, format='jpg', dpi=300, bbox_inches='tight')
plt.show()
print(f"Saved: {tsne_path}")

# Prediction grids
def denormalize(img_hwc):
    img = img_hwc.copy().astype(np.float32)
    img = (img - img.min()) / (img.max() - img.min() + 1e-8)
    return img

def save_visualization_grid(X, y_true, y_pred, y_probs, indices, title, filename):
    fig, axes = plt.subplots(1, 5, figsize=(16, 3.5), dpi=300)
    for ax, idx in zip(axes, indices):
        img    = denormalize(X[idx])
        pred   = y_pred[idx]
        true   = y_true[idx]
        conf   = y_probs[idx][pred]
        color  = 'green' if pred == true else 'red'
        status = '✓' if pred == true else '✗'
        ax.imshow(img)
        ax.axis('off')
        ax.set_title(
            f"{status} True: {LABEL_NAMES[true]}\n"
            f"  Pred: {LABEL_NAMES[pred]} ({conf:.0%})",
            color=color, fontsize=9, fontweight='bold', pad=4
        )
    fig.suptitle(title, fontsize=13, fontweight='bold', y=1.03)
    plt.tight_layout()
    save_path = os.path.join(SAVE_DIR, filename)
    plt.savefig(save_path, format='jpg', dpi=300, bbox_inches='tight')
    plt.show()
    print(f"Saved: {save_path}")

np.random.seed(42)
src_indices = np.random.choice(len(Xs), 5, replace=False)
tgt_indices = np.random.choice(len(Xt_eval), 5, replace=False)

print("\nSource domain predictions:")
save_visualization_grid(Xs, ys_int, src_preds_vis, src_probs_vis, src_indices,
                        "Wood Chip UDA — Source Domain (source_1)",
                        "source_predictions_grid.jpg")

print("\nTarget domain predictions:")
save_visualization_grid(Xt_eval, yt_int, tgt_preds_vis, tgt_probs_vis, tgt_indices,
                        "Wood Chip UDA — Target Domain (source_2)",
                        "target_predictions_grid.jpg")

# Copy sample images
src_sample_folder = os.path.join(SAVE_DIR, "source_samples")
tgt_sample_folder = os.path.join(SAVE_DIR, "target_samples")
os.makedirs(src_sample_folder, exist_ok=True)
os.makedirs(tgt_sample_folder, exist_ok=True)

def copy_original_images(paths, y_true, y_pred, y_probs, indices, folder, domain):
    print(f"\nCopying {domain} original images...")
    for rank, idx in enumerate(indices, start=1):
        true_label = LABEL_NAMES[y_true[idx]]
        pred_label = LABEL_NAMES[y_pred[idx]]
        conf       = y_probs[idx][y_pred[idx]]
        status     = 'Correct' if y_pred[idx] == y_true[idx] else 'Wrong'
        ext        = os.path.splitext(paths[idx])[1]
        fname      = f"{rank}_{status}_True-{true_label}_Pred-{pred_label}_Conf-{conf:.2f}{ext}"
        shutil.copy2(paths[idx], os.path.join(folder, fname))
        print(f"  {fname}")

copy_original_images(src_paths, ys_int, src_preds_vis, src_probs_vis,
                     src_indices, src_sample_folder, "Source")
copy_original_images(tgt_paths, yt_int, tgt_preds_vis, tgt_probs_vis,
                     tgt_indices, tgt_sample_folder, "Target")

src_correct = int(np.sum(src_preds_vis == ys_int))
tgt_correct = int(np.sum(tgt_preds_vis == yt_int))

print("\n" + "=" * 52)
print("PREDICTION SUMMARY")
print("=" * 52)
print(f"Source : {src_correct}/{len(ys_int)} correct ({accuracy_score(ys_int, src_preds_vis):.1%})")
print(f"Target : {tgt_correct}/{len(yt_int)} correct ({accuracy_score(yt_int, tgt_preds_vis):.1%})")
print("=" * 52)
print(f"\nAll files saved to: {SAVE_DIR}")
```
